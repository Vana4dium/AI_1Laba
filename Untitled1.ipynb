```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('all_ironman_suit.csv')
df.columns = df.columns.str.strip()

display(df.head())
df.info()

print(df.isnull().sum())

codename_mode = df['Codename'].mode()[0]
df['Codename'] = df['Codename'].fillna(codename_mode)

print(df.isnull().sum())

def parse_estimated_value(val):
    if pd.isna(val):
        return 0.0
    val = str(val).replace('$', '').strip()
    if val == '0':
        return 0.0
    if 'Billion' in val:
        return float(val.replace('Billion', '').strip()) * 1000.0
    if 'Million' in val:
        return float(val.replace('Million', '').strip())
    return float(val)

df['Estimated_Value_M'] = df['Estimated Value'].apply(parse_estimated_value)

scaler = MinMaxScaler()
df['Estimated_Value_Scaled'] = scaler.fit_transform(df[['Estimated_Value_M']])

display(df[['Model', 'Estimated Value', 'Estimated_Value_M', 'Estimated_Value_Scaled']].head(10))

df_final = pd.get_dummies(df, columns=['Era'], drop_first=True)

display(df_final.head())

df_final.to_csv("processed_ironman_suits.csv", index=False)
```
